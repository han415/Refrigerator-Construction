```python
import os
import pandas as pd
from ast import literal_eval

# 读取包含NER列的Excel文件
csv_file_path = 'dataset.csv'
df = pd.read_csv(csv_file_path)

# 假设NER列已经包含了单词列表，假设NER列的名称为'NER'
ner_column = 'NER'

# 将包含单词列表的字符串转换为实际的列表
df[ner_column] = df[ner_column].apply(literal_eval)

# 选择在全部数据中同时包含'cauliflower'和'cabbage'单词的前200条数据
both_cauliflower_and_cabbage_data = df[df[ner_column].apply(lambda x: 'cauliflower' in x and 'cabbage' in x)].head(200)

# 创建一个新的文件夹，如果已存在则不会引发错误
output_folder = 'selected_data'
os.makedirs(output_folder, exist_ok=True)

# 将选择的数据保存到新的Excel文件
output_excel_path = os.path.join(output_folder, 'both_cauliflower_and_cabbage.xlsx')
both_cauliflower_and_cabbage_data.to_excel(output_excel_path, index=False)

print(f'Selected data with both \'cauliflower\' and \'cabbage\' (first 200 rows) saved to {output_excel_path}')
```

```python
import os
import pandas as pd

# 定义文件夹路径
folder_path = 'selected_data'

# 获取文件夹中的所有Excel文件
excel_files = [f for f in os.listdir(folder_path) if f.endswith('.xlsx')]

# 创建一个空的DataFrame，用于存储合并后的数据
merged_data = pd.DataFrame()

# 逐个读取Excel文件并合并数据
for file in excel_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_excel(file_path)
    merged_data = merged_data.append(df, ignore_index=True)

# 删除重复的行
merged_data.drop_duplicates(inplace=True)

# 输出合并后的数据
print("Merged data without duplicates:")
print(merged_data)

# 将合并后的数据保存到新的Excel文件
output_merged_excel_path = 'data.xlsx'
merged_data.to_excel(output_merged_excel_path, index=False)

print(f'Merged data without duplicates saved to {output_merged_excel_path}')
```
